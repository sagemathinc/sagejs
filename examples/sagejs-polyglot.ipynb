# One mathematical workspace, six languages

Every cell below runs in the same Sage.js evaluator. Values are shared directly rather than serialized between language processes.

```sage
%%matlab
A = [1 2; 3 4];
size(A)
class(A)
```

```sage
%%sage
A[0, 0] = 9
R = QQ['t']
t = R.gen()
f = t^2 + 1
expression = sin(x)^2 + cos(x)^2
(A, f, expression)
```

```sage
%%magma
Type(A);
Type(f);
Factorization(2026);
A;
```

```sage
%%maple
whattype(A);
whattype(f);
factor(2026);
```

```sage
%%wolfram
Head[A]
Dimensions[A]
Head[f]
FactorInteger[2026]
```

```sage
%%python
print(A)
print(f)
A[0, 1] = 77
```

```sage
%%sage
plot(sin(x^2), (x, 0, 2*pi),
     title='Shared Sage.js workspace')
```
